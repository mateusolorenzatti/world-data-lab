```python
import os, sys
from datetime import datetime

import kagglehub

sys.path.append(os.path.join(os.getcwd(), "/home/jovyan/work/src/core"))

import spark_session
import pyspark.sql.functions as F
```

```python
spark = spark_session.build_spark()
```

```python
path = kagglehub.dataset_download("tawfikelmetwally/automobile-dataset")
path = path + '/Automobile.csv'
```

```python
cardt_df = spark.read.option("header", "true") \
                            .csv(path)
```

```python
cardt_df = cardt_df.withColumn('ingestion_datetime', F.lit(datetime.now())) \
                    .withColumn('ingestion_file',  F.lit(path))
```

```python
spark_session.run_sql('CREATE SCHEMA IF NOT EXISTS bro_cardt;')
```

```python
spark_session.write_table(cardt_df, 'bro_cardt', 'automobile', 'append')
```
