### [GENDT] Automobile (Group By Country) - Gold Layer

Import libs and start spark context

```python
import os, sys

sys.path.append(os.path.join(os.getcwd(), "/home/jovyan/work/src/core"))

import spark_session
import pyspark.sql.functions as F
from pyspark.sql import Window
```

```python
spark = spark_session.build_spark()
```

Start variables and read table

```python
origin_table = 'svr_gendt.automobile'

target_schema = 'gld_gendt'
target_table = 'automobile_by_country'
```

```python
df_automobile = spark_session.read_table(spark, origin_table)
```

Processing and grouping data

```python
df_automobile = df_automobile.groupBy("s_country_origin", "d_slv_processing").agg(
    F.round(F.avg("v_miles_per_galoon"), 2).alias("v_avg_miles_per_galoon"),
    F.round(F.max("v_miles_per_galoon"), 2).alias("v_max_miles_per_galoon"),
    F.round(F.avg("v_cylinders"), 2).alias("v_avg_cylinders"),
    F.round(F.avg("v_horsepower"), 2).alias("v_avg_horsepower"),
    F.round(F.avg("v_weight"), 2).alias("v_avg_weight"),
    F.round(F.avg("v_acceleration"), 2).alias("v_avg_acceleration")
)
```

Structure dataframe and write

```python
spark_session.run_sql(f'CREATE SCHEMA IF NOT EXISTS {target_schema};')
```

```python
spark_session.write_table(df_automobile, target_schema, target_table, 'overwrite')
```

```python
spark.stop()
```
