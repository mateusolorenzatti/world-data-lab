### [GENDT] Automobile - Bronze Layer

Import libs and start spark context

```python
import os, sys
from datetime import datetime

import kagglehub

sys.path.append(os.path.join(os.getcwd(), "/home/jovyan/work/src/core"))

import spark_session
import pyspark.sql.functions as F
```

```python
spark = spark_session.build_spark()
```

Start variables and ingest file

```python
target_schema = 'brz_gendt'
target_table = 'automobile'
```

```python
path = kagglehub.dataset_download("tawfikelmetwally/automobile-dataset")
path = path + '/Automobile.csv'
```

```python
df_automobile = spark.read.option("header", "true") \
                            .csv(path)
```

Add control fields and write

```python
df_automobile = df_automobile.withColumn('ingestion_datetime', F.lit(datetime.now())) \
                            .withColumn('ingestion_file',  F.lit(path))
```

```python
spark_session.run_sql(f'CREATE SCHEMA IF NOT EXISTS {target_schema};')
```

```python
spark_session.write_table(df_automobile, target_schema, target_table, 'append')
```

```python
spark.stop()
```
